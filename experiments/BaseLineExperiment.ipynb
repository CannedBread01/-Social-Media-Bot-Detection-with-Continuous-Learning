#### Base Line Experiment Setup
- test pipeline performance with the base classifier on each dataset individually

```python
# imports
import torch
from dataset_utils import *
from pipeline_utils.ClassificationHeads import BaselineClassificationHead
from pipeline_utils.pipeline_utils import test_classifier
from sklearn.metrics import accuracy_score, recall_score, f1_score, confusion_matrix
from pipeline_utils.pipeline_utils import train_classifier
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
def train_loop(data_loader, classifier):
    ground_truth_labels = []
    embedding_features = []
    for i, sample in enumerate(data_loader):
        ground_truth_labels.append(sample[1])
        embedding_features.append(sample[0])
        print(f"\rIteration: {i}", end="")

    # perform classification training
    criterion = torch.nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(classifier.parameters(), lr=0.01)
    ground_truths, predictions = train_classifier(classifier, criterion, optimizer, embedding_features, ground_truth_labels, epochs=10)

    # performance
    print("")
    print("[training performance]")
    print("accuracy: ", accuracy_score(ground_truths, predictions))
    print("recall: ", recall_score(ground_truths, predictions, average="macro"))
    print("f1 score: ", f1_score(ground_truths, predictions, average="macro"))
    print(confusion_matrix(ground_truths, predictions))
    return predictions, ground_truths

def testing_loop(data_loader, classifier):
    ground_truth_labels = []
    embedding_features = []
    for i, sample in enumerate(data_loader):
        ground_truth_labels.append(sample[1])
        embedding_features.append(sample[0])
        print(f"\rIteration: {i}", end="")

    # perform classification training
    ground_truths, predictions = test_classifier(classifier, embedding_features, ground_truth_labels)

    # performance
    print("")
    print("[testing performance]")
    print("accuracy: ", accuracy_score(ground_truths, predictions))
    print("recall: ", recall_score(ground_truths, predictions, average="macro"))
    print("f1 score: ", f1_score(ground_truths, predictions, average="macro"))
    print(confusion_matrix(ground_truths, predictions))
    return predictions, ground_truths
```

```python
# Dataset: Cresci-2017

# init relevant models
classifier = BaselineClassificationHead(45,6).to(device)
# train loop
dataset = ProcessedDataset("train", "../datasets/ProcessedDatasets/", "Cresci17")
train_loop(dataset, classifier)
# test loop
dataset = ProcessedDataset("test", "../datasets/ProcessedDatasets/", "Cresci17")
testing_loop(dataset, classifier)
print("--done--")
```
